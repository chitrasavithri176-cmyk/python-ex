```python
import pandas as pd

# Load the CSV file into a DataFrame
df = pd.read_csv('/content/global_top2000.csv')

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Sort the DataFrame by 'Subscribers' in descending order to find top creators
top_creators = df.sort_values(by='Subscribers', ascending=False)

# Display the top 10 creators
display(top_creators.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a histogram of subscriber counts
plt.figure(figsize=(12, 6))
sns.histplot(df['Subscribers'], bins=50, kde=True)
plt.title('Distribution of Subscriber Counts')
plt.xlabel('Subscribers')
plt.ylabel('Number of Channels')
plt.xscale('log') # Use a log scale for better visualization of skewed data
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a histogram of total views
plt.figure(figsize=(12, 6))
sns.histplot(df['Total Views'], bins=50, kde=True)
plt.title('Distribution of Total Views')
plt.xlabel('Total Views')
plt.ylabel('Number of Channels')
plt.xscale('log') # Use a log scale for better visualization of skewed data
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Analyze the number of channels per country
country_counts = df['Country'].value_counts().reset_index()
country_counts.columns = ['Country', 'Number of Channels']

# Display the top 10 countries by number of channels
display(country_counts.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the top 10 countries by number of channels
plt.figure(figsize=(14, 7))
sns.barplot(x='Country', y='Number of Channels', data=country_counts.head(10), palette='viridis')
plt.title('Top 10 Countries by Number of Channels')
plt.xlabel('Country')
plt.ylabel('Number of Channels')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the top 10 countries by number of channels
plt.figure(figsize=(14, 7))
sns.barplot(x='Country', y='Number of Channels', data=country_counts.head(10), palette='viridis', hue='Country', legend=False)
plt.title('Top 10 Countries by Number of Channels')
plt.xlabel('Country')
plt.ylabel('Number of Channels')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Calculate the average subscribers by country
average_subscribers_by_country = df.groupby('Country')['Subscribers'].mean().reset_index()
average_subscribers_by_country.columns = ['Country', 'Average Subscribers']

# Sort in descending order to see the top countries by average subscribers
average_subscribers_by_country = average_subscribers_by_country.sort_values(by='Average Subscribers', ascending=False)

# Display the top 10 countries by average subscribers
display(average_subscribers_by_country.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the top 10 countries by average subscribers
plt.figure(figsize=(14, 7))
sns.barplot(x='Country', y='Average Subscribers', data=average_subscribers_by_country.head(10), palette='magma')
plt.title('Top 10 Countries by Average Subscribers')
plt.xlabel('Country')
plt.ylabel('Average Subscribers')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Calculate the correlation between 'Subscribers' and 'Total Videos'
correlation = df['Subscribers'].corr(df['Total Videos'])

# Display the correlation
print(f"Correlation between Subscribers and Total Videos: {correlation}")
```
