```python
from abc import ABC,abstractmethod
import math
class shape(ABC):
  @abstractmethod
  def area(self):
    pass
  @abstractmethod
  def perimeter(self):
    pass
class circle(shape):
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return math.pi*self.radius**2
  def perimeter(self):
    return 2*math.pi*self.radius
class rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    return self.length*self.width
  def perimeter(self):
    return 2*(self.length+self.width)
class triangle(shape):
  def __init__(self,base,height,a,b,c):
    self.base=base
    self.height=height
    self.a,self.b,self.c= a,b,c
  def area(self):
    return 0.5*self.base*self.height
  def perimeter(self):
    return self.a+self.b+self.c
shapes = [circle(5), rectangle(4,6), triangle(6,4,5,5,6)]
for shape in shapes:
  print(f"{shape.__class__.__name__}-> area:{shape.area():.2f}, perimeter:{shape.perimeter():.2f}")
```

```python
from abc import ABC, abstractmethod


# ---------- Custom Exceptions ----------
class ItemNotFoundError(Exception):
    """Raised when a requested item does not exist in the library."""
    pass


class ItemAlreadyIssuedError(Exception):
    """Raised when trying to issue an item that is already issued."""
    pass


# ---------- Abstraction ----------
class Media(ABC):
    """Abstract base class representing any item the library can hold."""

    def __init__(self, title, item_id):
        self._title = title            # protected attribute (encapsulation)
        self.__item_id = item_id       # private attribute (encapsulation)
        self.__is_issued = False       # private attribute

    # ---- Encapsulation: controlled access via getters/setters ----
    @property
    def item_id(self):
        return self.__item_id

    @property
    def is_issued(self):
        return self.__is_issued

    def issue(self):
        if self.__is_issued:
            raise ItemAlreadyIssuedError(f"'{self._title}' is already issued.")
        self.__is_issued = True

    def return_item(self):
        self.__is_issued = False

    @abstractmethod
    def display_info(self):
        """Every subclass MUST implement its own version (abstraction)."""
        pass


# ---------- Inheritance + Polymorphism ----------
class Book(Media):
    def __init__(self, title, item_id, author, pages):
        super().__init__(title, item_id)
        self.author = author
        self.pages = pages

    def display_info(self):   # polymorphic override
        status = "Issued" if self.is_issued else "Available"
        return f"[Book]  {self._title} by {self.author} ({self.pages} pages) - {status}"


class DVD(Media):
    def __init__(self, title, item_id, director, duration_minutes):
        super().__init__(title, item_id)
        self.director = director
        self.duration_minutes = duration_minutes

    def display_info(self):   # polymorphic override
        status = "Issued" if self.is_issued else "Available"
        return f"[DVD]   {self._title} directed by {self.director} ({self.duration_minutes} min) - {status}"


# ---------- Library (uses data structures + file handling) ----------
class Library:
    def __init__(self):
        self.catalog = {}   # dictionary: item_id -> Media object

    def add_item(self, media_item):
        self.catalog[media_item.item_id] = media_item

    def find_item(self, item_id):
        if item_id not in self.catalog:
            raise ItemNotFoundError(f"No item found with ID '{item_id}'.")
        return self.catalog[item_id]

    def issue_item(self, item_id):
        item = self.find_item(item_id)
        item.issue()
        print(f"Issued: {item._title}")

    def return_item(self, item_id):
        item = self.find_item(item_id)
        item.return_item()
        print(f"Returned: {item._title}")

    def show_catalog(self):
        print("\n--- LIBRARY CATALOG ---")
        for item in self.catalog.values():   # list-like iteration over dict values
            print(item.display_info())

    # ---------- File Handling ----------
    def save_to_file(self, filename="catalog.txt"):
        try:
            with open(filename, "w") as f:
                for item in self.catalog.values():
                    f.write(item.display_info() + "\n")
            print(f"\nCatalog saved to '{filename}'.")
        except IOError as e:
            print(f"Error saving file: {e}")

    def load_from_file(self, filename="catalog.txt"):
        try:
            with open(filename, "r") as f:
                print(f"\n--- CONTENTS OF '{filename}' ---")
                print(f.read())
        except FileNotFoundError:
            print(f"Error: '{filename}' does not exist yet. Save the catalog first.")


# ---------- Demo / Driver Code ----------
def main():
    library = Library()
    library.add_item(Book("The Alchemist", "B001", "Paulo Coelho", 197))
    library.add_item(Book("Clean Code", "B002", "Robert C. Martin", 464))
    library.add_item(DVD("Inception", "D001", "Christopher Nolan", 148))

    library.show_catalog()

    # Exception handling in action
    try:
        library.issue_item("B001")
        library.issue_item("B001")   # will raise ItemAlreadyIssuedError
    except ItemAlreadyIssuedError as e:
        print(f"\nError: {e}")

    try:
        library.issue_item("X999")   # will raise ItemNotFoundError
    except ItemNotFoundError as e:
        print(f"Error: {e}")

    library.return_item("B001")
    library.show_catalog()

    library.save_to_file()
    library.load_from_file()


if __name__ == "__main__":
    main()
```

```python
from abc import ABC, abstractmethod
from datetime import datetime


# ---------- Custom Exceptions ----------
class InvalidAmountError(Exception):
    """Raised when a deposit/withdrawal amount is not valid."""
    pass


class InsufficientFundsError(Exception):
    """Raised when a withdrawal exceeds the available balance."""
    pass


class AccountNotFoundError(Exception):
    """Raised when an account number does not exist in the bank."""
    pass


# ---------- Abstraction + Encapsulation ----------
class Account(ABC):
    def __init__(self, account_number, holder_name, balance=0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.__balance = balance          # private attribute
        self.transactions = []            # list (data structure) of transaction strings

    # controlled read-only access to a private attribute
    @property
    def balance(self):
        return self.__balance

    def _log(self, message):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        self.transactions.append(f"[{timestamp}] {message}")

    def deposit(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Deposit amount must be positive.")
        self.__balance += amount
        self._log(f"Deposited {amount:.2f}. New balance: {self.__balance:.2f}")

    def withdraw(self, amount):
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be positive.")
        if amount > self.__balance:
            raise InsufficientFundsError(
                f"Cannot withdraw {amount:.2f}; balance is only {self.__balance:.2f}."
            )
        self.__balance -= amount
        self._log(f"Withdrew {amount:.2f}. New balance: {self.__balance:.2f}")

    @abstractmethod
    def account_type(self):
        """Forces every subclass to identify itself (abstraction)."""
        pass


# ---------- Inheritance + Polymorphism ----------
class SavingsAccount(Account):
    INTEREST_RATE = 0.04   # 4% per annum

    def account_type(self):
        return "Savings Account"

    def apply_interest(self):
        interest = self.balance * self.INTEREST_RATE
        self.deposit(interest)
        self._log(f"Interest of {interest:.2f} applied.")


class CurrentAccount(Account):
    OVERDRAFT_LIMIT = 5000

    def account_type(self):
        return "Current Account"

    def withdraw(self, amount):   # polymorphic override: allows overdraft
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be positive.")
        if amount > self.balance + self.OVERDRAFT_LIMIT:
            raise InsufficientFundsError(
                f"Withdrawal exceeds overdraft limit of {self.OVERDRAFT_LIMIT}."
            )
        # Overdraft allowed: adjust the private balance directly (name-mangled attribute)
        new_balance = self.balance - amount
        self._Account__balance = new_balance
        self._log(f"Withdrew {amount:.2f} (overdraft allowed). New balance: {new_balance:.2f}")


# ---------- Bank (data structures + file handling) ----------
class Bank:
    def __init__(self):
        self.accounts = {}   # dictionary: account_number -> Account object

    def open_account(self, account):
        self.accounts[account.account_number] = account
        print(f"Opened {account.account_type()} for {account.holder_name} "
              f"(A/C: {account.account_number})")

    def find_account(self, account_number):
        if account_number not in self.accounts:
            raise AccountNotFoundError(f"Account '{account_number}' not found.")
        return self.accounts[account_number]

    def print_statement(self, account_number):
        account = self.find_account(account_number)
        print(f"\n--- Statement for {account.holder_name} ({account.account_type()}) ---")
        for entry in account.transactions:
            print(entry)
        print(f"Current Balance: {account.balance:.2f}")

    # ---------- File Handling ----------
    def export_all_transactions(self, filename="transactions.txt"):
        try:
            with open(filename, "w") as f:
                for account in self.accounts.values():
                    f.write(f"=== {account.holder_name} ({account.account_number}) ===\n")
                    for entry in account.transactions:
                        f.write(entry + "\n")
                    f.write(f"Final Balance: {account.balance:.2f}\n\n")
            print(f"\nAll transactions exported to '{filename}'.")
        except IOError as e:
            print(f"File error: {e}")

    def read_transaction_log(self, filename="transactions.txt"):
        try:
            with open(filename, "r") as f:
                print(f"\n--- READING '{filename}' ---")
                print(f.read())
        except FileNotFoundError:
            print(f"'{filename}' not found. Export transactions first.")


# ---------- Demo / Driver Code ----------
def main():
    bank = Bank()

    savings = SavingsAccount("SB001", "Ananya Rao", 10000)
    current = CurrentAccount("CA001", "Rahul Mehta", 2000)

    bank.open_account(savings)
    bank.open_account(current)

    try:
        savings.deposit(5000)
        savings.apply_interest()
        current.withdraw(6000)   # uses overdraft, still succeeds
        savings.withdraw(2000)

        # This will raise InsufficientFundsError
        savings.withdraw(1_000_000)
    except InsufficientFundsError as e:
        print(f"\nTransaction failed: {e}")
    except InvalidAmountError as e:
        print(f"\nInvalid input: {e}")

    try:
        bank.find_account("ZZ999")
    except AccountNotFoundError as e:
        print(f"Error: {e}")

    bank.print_statement("SB001")
    bank.print_statement("CA001")

    bank.export_all_transactions()
    bank.read_transaction_log()


if __name__ == "__main__":
    main()

```

```python
import csv
from abc import ABC, abstractmethod


# ---------- Custom Exceptions ----------
class InvalidMarksError(Exception):
    """Raised when a mark is outside the valid 0-100 range."""
    pass


# ---------- Abstraction + Inheritance + Polymorphism ----------
class Grader(ABC):
    """Base class for different grading policies."""

    @abstractmethod
    def get_grade(self, average):
        pass


class StandardGrader(Grader):
    def get_grade(self, average):
        if average >= 90: return "A+"
        elif average >= 80: return "A"
        elif average >= 70: return "B"
        elif average >= 60: return "C"
        elif average >= 50: return "D"
        else: return "F"


class StrictGrader(Grader):   # polymorphic alternative grading policy
    def get_grade(self, average):
        if average >= 95: return "A+"
        elif average >= 85: return "A"
        elif average >= 75: return "B"
        elif average >= 65: return "C"
        elif average >= 55: return "D"
        else: return "F"


# ---------- Encapsulation ----------
class Student(object):
    def __init__(self, name, roll_no, marks):
        self._name = name
        self._roll_no = roll_no
        self.__marks = self._validate_marks(marks)   # private, validated data

    @staticmethod
    def _validate_marks(marks):
        for subject, score in marks.items():
            if not (0 <= score <= 100):
                raise InvalidMarksError(
                    f"Invalid mark for {subject}: {score}. Must be between 0 and 100."
                )
        return marks

    @property
    def name(self):
        return self._name

    @property
    def roll_no(self):
        return self._roll_no

    @property
    def marks(self):
        return self.__marks

    def average(self):
        return sum(self.__marks.values()) / len(self.__marks)


# ---------- Functions working on the data structure (list of Student objects) ----------
def add_student(students, name, roll_no, marks):
    """Adds a new Student to the list. Raises InvalidMarksError if marks are bad."""
    student = Student(name, roll_no, marks)
    students.append(student)
    return student


def calculate_results(students, grader: Grader):
    """Returns a list of dicts summarizing each student's result (uses polymorphic grader)."""
    results = []
    for s in students:
        avg = s.average()
        results.append({
            "roll_no": s.roll_no,
            "name": s.name,
            "average": round(avg, 2),
            "grade": grader.get_grade(avg),
        })
    return results


def display_results(results):
    print(f"{'Roll No':<10}{'Name':<15}{'Average':<10}{'Grade':<5}")
    print("-" * 40)
    for r in results:
        print(f"{r['roll_no']:<10}{r['name']:<15}{r['average']:<10}{r['grade']:<5}")


# ---------- File Handling (CSV) ----------
def save_results_to_csv(results, filename="results.csv"):
    try:
        with open(filename, "w", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=["roll_no", "name", "average", "grade"])
            writer.writeheader()
            writer.writerows(results)
        print(f"\nResults saved to '{filename}'.")
    except IOError as e:
        print(f"Error writing file: {e}")


def load_results_from_csv(filename="results.csv"):
    try:
        with open(filename, "r") as f:
            reader = csv.DictReader(f)
            print(f"\n--- CONTENTS OF '{filename}' ---")
            for row in reader:
                print(row)
    except FileNotFoundError:
        print(f"Error: '{filename}' not found. Save results first.")


# ---------- Demo / Driver Code ----------
def main():
    students = []

    # Adding students; wrapped in try/except for exception handling
    try:
        add_student(students, "Priya Sharma", "R001",
                    {"Math": 92, "Science": 88, "English": 79})
        add_student(students, "Karan Singh", "R002",
                    {"Math": 65, "Science": 70, "English": 60})
        add_student(students, "Meera Nair", "R003",
                    {"Math": 45, "Science": 50, "English": 40})

        # This will raise InvalidMarksError (150 is out of range)
        add_student(students, "Bad Data", "R004", {"Math": 150})
    except InvalidMarksError as e:
        print(f"Skipping invalid record: {e}")

    print("\n=== Results using StandardGrader ===")
    standard_results = calculate_results(students, StandardGrader())
    display_results(standard_results)

    print("\n=== Results using StrictGrader (polymorphism: different grading policy) ===")
    strict_results = calculate_results(students, StrictGrader())
    display_results(strict_results)

    save_results_to_csv(standard_results)
    load_results_from_csv()


if __name__ == "__main__":
    main()
```
