### Delivery Time Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Descriptive statistics for Swiggy and Zomato average delivery times
print("Descriptive Statistics for Swiggy Average Delivery Time (minutes):")
display(df['swiggy_avg_delivery_time_minutes'].describe())

print("\nDescriptive Statistics for Zomato Average Delivery Time (minutes):")
display(df['zomato_avg_delivery_time_minutes'].describe())

# Plotting the distribution of Swiggy average delivery time
fig, axes = plt.subplots(1, 2, figsize=(20, 6))
sns.histplot(df['swiggy_avg_delivery_time_minutes'], kde=True, bins=30, color='skyblue', ax=axes[0])
axes[0].set_title('Distribution of Swiggy Average Delivery Time (minutes)')
axes[0].set_xlabel('Swiggy Average Delivery Time (minutes)')
axes[0].set_ylabel('Frequency')
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Plotting the distribution of Zomato average delivery time
sns.histplot(df['zomato_avg_delivery_time_minutes'], kde=True, bins=30, color='salmon', ax=axes[1])
axes[1].set_title('Distribution of Zomato Average Delivery Time (minutes)')
axes[1].set_xlabel('Zomato Average Delivery Time (minutes)')
axes[1].set_ylabel('Frequency')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Let's also look at how delivery time relates to the distance from the city center.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Scatter plot of delivery times vs. distance_from_city_center_km
fig = plt.figure(figsize=(12, 7))
sns.scatterplot(x='distance_from_city_center_km', y='swiggy_avg_delivery_time_minutes', data=df, alpha=0.6, label='Swiggy Delivery Time', color='skyblue')
sns.scatterplot(x='distance_from_city_center_km', y='zomato_avg_delivery_time_minutes', data=df, alpha=0.6, label='Zomato Delivery Time', color='salmon')
plt.title('Average Delivery Time vs. Distance From City Center for Swiggy and Zomato')
plt.xlabel('Distance From City Center (km)')
plt.ylabel('Average Delivery Time (minutes)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Discount Frequency Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Descriptive statistics for Swiggy and Zomato discount frequency percentages
print("Descriptive Statistics for Swiggy Discount Frequency Percentage:")
display(df['swiggy_discount_frequency_pct'].describe())

print("\nDescriptive Statistics for Zomato Discount Frequency Percentage:")
display(df['zomato_discount_frequency_pct'].describe())

# Plotting the distribution of Swiggy and Zomato discount frequency percentages
fig, axes = plt.subplots(1, 2, figsize=(20, 6))

sns.histplot(df['swiggy_discount_frequency_pct'], kde=True, bins=20, color='skyblue', ax=axes[0])
axes[0].set_title('Distribution of Swiggy Discount Frequency (%)')
axes[0].set_xlabel('Discount Frequency Percentage')
axes[0].set_ylabel('Frequency')
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

sns.histplot(df['zomato_discount_frequency_pct'], kde=True, bins=20, color='salmon', ax=axes[1])
axes[1].set_title('Distribution of Zomato Discount Frequency (%)')
axes[1].set_xlabel('Discount Frequency Percentage')
axes[1].set_ylabel('Frequency')
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Let's also visualize the distribution of discount frequencies using box plots for a more direct comparison of their central tendency and spread.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for a combined box plot
discount_data = pd.DataFrame({
    'Platform': ['Swiggy'] * len(df) + ['Zomato'] * len(df),
    'Discount_Frequency_Pct': pd.concat([
        df['swiggy_discount_frequency_pct'],
        df['zomato_discount_frequency_pct']
    ])
})

fig = plt.figure(figsize=(10, 6))
sns.boxplot(x='Platform', y='Discount_Frequency_Pct', data=discount_data, palette={'Swiggy': 'skyblue', 'Zomato': 'salmon'})
plt.title('Comparison of Discount Frequency Percentage between Swiggy and Zomato')
plt.xlabel('Platform')
plt.ylabel('Discount Frequency Percentage')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Estimated Monthly Revenue Comparison

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Descriptive statistics for Swiggy and Zomato estimated monthly revenue
print("Descriptive Statistics for Swiggy Estimated Monthly Revenue (INR):")
display(df['swiggy_estimated_monthly_revenue_inr'].describe())

print("\nDescriptive Statistics for Zomato Estimated Monthly Revenue (INR):")
display(df['zomato_estimated_monthly_revenue_inr'].describe())

# Plotting the distribution of Swiggy and Zomato estimated monthly revenue
fig, axes = plt.subplots(1, 2, figsize=(20, 6))

sns.histplot(df['swiggy_estimated_monthly_revenue_inr'], kde=True, bins=30, color='skyblue', ax=axes[0])
axes[0].set_title('Distribution of Swiggy Estimated Monthly Revenue (INR)')
axes[0].set_xlabel('Estimated Monthly Revenue (INR)')
axes[0].set_ylabel('Frequency')
axes[0].ticklabel_format(style='plain', axis='x') # Prevent scientific notation on x-axis
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

sns.histplot(df['zomato_estimated_monthly_revenue_inr'], kde=True, bins=30, color='salmon', ax=axes[1])
axes[1].set_title('Distribution of Zomato Estimated Monthly Revenue (INR)')
axes[1].set_xlabel('Estimated Monthly Revenue (INR)')
axes[1].set_ylabel('Frequency')
axes[1].ticklabel_format(style='plain', axis='x') # Prevent scientific notation on x-axis
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

Let's also visualize the overall average monthly revenue for each platform with a bar chart for a quick comparison.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average monthly revenue for each platform
avg_revenue_swiggy = df['swiggy_estimated_monthly_revenue_inr'].mean()
avg_revenue_zomato = df['zomato_estimated_monthly_revenue_inr'].mean()

# Create a DataFrame for plotting
revenue_data = pd.DataFrame({
    'Platform': ['Swiggy', 'Zomato'],
    'Average_Monthly_Revenue_INR': [avg_revenue_swiggy, avg_revenue_zomato]
})

fig = plt.figure(figsize=(8, 6))
sns.barplot(x='Platform', y='Average_Monthly_Revenue_INR', data=revenue_data, palette={'Swiggy': 'skyblue', 'Zomato': 'salmon'})
plt.title('Average Estimated Monthly Revenue: Swiggy vs Zomato')
plt.xlabel('Platform')
plt.ylabel('Average Monthly Revenue (INR)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ticklabel_format(style='plain', axis='y') # Prevent scientific notation on y-axis
plt.show()
```
