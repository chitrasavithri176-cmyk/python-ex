```python
# how many products are Amezon's choice

print(df["is_amazon_choice"].value_counts())

choice=df["is_amazon_choice"].value_counts()
plt.figure(figsize=(6,5))
choice.plot(kind="pie",autopct="%1.1f%%")
plt.title("Amazon's Choice")
plt.ylabel("")
plt.show()
```

```python
# find best seller products

result=df[df["is_best_seller"]==True]
print(result[["product_title","category","price_inr","rating"]])
```

```python
# Find products that are both Best sellers and Amazon's choice

result=df[(df["is_best_seller"]==True) & (df["is_amazon_choice"]==True)]
print(result)
```

```python
# Average price of Best seller

result=df[df["is_best_seller"]==True]
print("Average price of Best seller:",result["price_inr"].mean())
```

```python
# Calculate the amount saved on each products

df["amount_saved"]=(df["original_price_inr"]-df["price_inr"])
print(df[["product_title","price_inr","original_price_inr","amount_saved"]].head())
```

```python
# Final products summary by category

summary=df.groupby("category").agg(
    total_products=("asin","count"),
    average_price=("price_inr","mean"),
    average_ratings=("rating","mean"),
    average_reviews=("num_ratings","mean"),
    highest_price=("price_inr","max"),
    lowest_price=("price_inr","min")
).reset_index()
print(summary)
```

```python
# Save the final pandas analysis to CSV

summary.to_csv("Amazon_Product_Category_summary.csv",index=False)
print("project completed successfully!")
print("Output file:Amazon_product_Category_Summary.csv")
```

## Business Insight and Summary

The dataset contains 623 unique products across 12 categories, with gaming console being the most frequent category.
The average product price is approximately 28,231 INR, with prices ranging from 139 INR to 1,117,486 INR.
Television, camera, and laptop categories have the highest average prices, while earbuds, headphones, and gaming console have the lowest.
The average discount across all products is 40.95%, with smartwatch, earbuds, and headphones categories showing the highest average discounts.
No products were identified as Prime in the dataset, and similarly, no products were marked as best sellers.
11 products were identified as Amazon's Choice.
The rating distribution shows a concentration of products with ratings around 3.8 to 4.2.
The discount distribution indicates a significant number of products with discounts in the 30-60% range.
A detailed summary by category, including total products, average price, ratings, reviews, highest, and lowest prices, has been generated and saved to Amazon_Product_Category_summary.csv.
This analysis offers a comprehensive overview of the product landscape, pricing strategies, and customer sentiment within the dataset.
